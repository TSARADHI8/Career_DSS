```python
df = pd.read_csv("placements_featured.csv")
```

```python
#

def job_score(row):
    score = 0
    if row["Placement_Readiness"] == "High":
        score += 2
    if row["Experience_Level"] in ["High", "Medium"]:
        score += 1
    if row["Academic_Risk"] != "High":
        score += 1
    return score
```

```python
def higher_studies_score(row):
    score = 0
    if row["Higher Studies"] == "Yes":
        score += 2
    if row["Academic_Risk"] == "Low":
        score += 1
    if row["Experience_Level"] in ["High", "Medium"]:
        score += 1
    return score
```

```python
def skill_improvement_score(row):
    score = 0
    if row["Placement_Readiness"] == "Medium":
        score += 2
    if row["Skill_Count"] < 3:
        score += 1
    if row["Academic_Risk"] != "High":
        score += 1
    return score
```

```python
def wait_improve_score(row):
    score = 0
    if row["Placement_Readiness"] == "Low":
        score += 2
    if row["Academic_Risk"] == "High":
        score += 1
    return score
```

```python
# Apply DSS Logic to Each Student

df["Job_Score"] = df.apply(job_score, axis=1)
```

```python
df["Higher_Studies_Score"] = df.apply(higher_studies_score, axis=1)
```

```python
df["Skill_Improvement_Score"] = df.apply(skill_improvement_score, axis=1)
```

```python
df["Wait_Improve_Score"] = df.apply(wait_improve_score, axis=1)
```

```python
# Identify Top Career Options

def recommend(row):
    scores = {
        "Immediate Job": row["Job_Score"],
        "Higher Studies": row["Higher_Studies_Score"],
        "Skill Improvement": row["Skill_Improvement_Score"],
        "Wait & Improve": row["Wait_Improve_Score"]
    }
    return sorted(scores, key=scores.get, reverse=True)[:2]

df["Recommended_Options"] = df.apply(recommend, axis=1)
```

```python
df.to_csv("placements_dss_output.csv", index=False)

print("STEP-7 completed: DSS recommendations generated")
```
