```python
import pandas as pd
```

```python
df= pd.read_csv("placements_cleaned.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
# Feature engineering 

# creating a new column called Experinece level using intenrships and projects 

def get_experience_level(row):
    if row["Internships"] >= 2 and row["Projects Completed"] >= 5:
        return "High"
    elif row["Internships"] >= 1 or row["Projects Completed"] >= 3:
        return "Medium"
    else:
        return "Low"

df["Experience_Level"] = df.apply(get_experience_level, axis=1)
```

```python
# creating new column called Academic_risk based on the backlogs

def academic_risk(backlogs):
    if backlogs <= 1:
        return "Low"
    elif backlogs <= 3:
        return "Medium"
    else:
        return "High"

df["Academic_Risk"] = df["Backlogs"].apply(academic_risk)
```

```python
# creating a new column called Placement Readiness 

df.columns
```

```python
# creating new coulumn called skill_Count

df["Skill_Count"] = df["Skills"].apply(
    lambda x: len(str(x).split(",")) if pd.notna(x) else 0
)
```

```python
df.columns
```

```python
# creating new column called placement readiness using skill_count, experience_level and academic_risk 

def placement_readiness(row):
    if row["Skill_Count"] >= 3 and row["Experience_Level"] == "High" and row["Academic_Risk"] == "Low":
        return "High"
    elif row["Skill_Count"] >= 2 and row["Academic_Risk"] != "High":
        return "Medium"
    else:
        return "Low"

df["Placement_Readiness"] = df.apply(placement_readiness, axis=1)
```
